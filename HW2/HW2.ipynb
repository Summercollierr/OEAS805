```python
import matplotlib.pyplot as plt
import pandas as pd

#Clean 'Cigarette Butts' column
data["Cigarette Butts Cleaned"] = pd.to_numeric(
    data["Cigarette Butts"].astype(str) .str.replace(r"[^\d.]", "", regex=True), errors="coerce",).fillna(0)

#Get ALL areas sorted from lowest to highest count
all_cigs = ( data.groupby("Area")["Cigarette Butts Cleaned"].sum() .sort_values())

# Calculate dynamic height (0.25 inches per location, minimum 6 inches)
num_locations = len(all_cigs)
fig_height = max(6, num_locations * 0.25)

# Draw the lollipop chart
plt.figure(figsize=(9, fig_height))
plt.hlines(
    y=all_cigs.index,
    xmin=0,
    xmax=all_cigs.values,
    color="lightcoral",
    linewidth=1.5,
)
plt.plot(all_cigs.values, all_cigs.index, "o", color="firebrick", markersize=6)


plt.title(
    f"Cigarette Butts Collected Across All {num_locations} Locations",
    fontweight="bold",
    pad=15,
)
plt.xlabel("Total Cigarette Butts")
plt.gca().xaxis.set_major_formatter("{x:,.0f}")
plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

I chose to use the lollipop chart because it is easy to read. I will attempt to use more. 

```python
# List of smoking-related columns
smoking_cols = [
    'Cigarette Butts', 
    'Cigar Tips', 
    'Tobacco Packaging/Wrap', 
    'Cigarette Lighters'
]

# Clean values
for col in smoking_cols:
    data[col] = data[col].astype(str).str.replace(',', '').str.strip()
    data[col] = pd.to_numeric(data[col], errors='coerce').fillna(0)

# Sum up totals
smoking_totals = data[smoking_cols].sum()

# Plot simple pie chart
plt.figure(figsize=(6, 6))
plt.pie(
    smoking_totals, 
    labels=smoking_totals.index, 
    autopct='%1.1f%%', 
    colors=['#d95f02', '#7570b3', '#e7298a', '#66a61e'],
    startangle=140
)
plt.title('Breakdown of Smoking-Related Litter')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
data = pd.read_csv('/Users/summercollier/OEAS805/HW2/International_Coastal_Cleanup_20260908.csv')

# Clean both columns (remove commas and convert to numbers)
data['Beverage Bottles (Plastic)'] = data['Beverage Bottles (Plastic)'].astype(str).str.replace(',', '').str.strip()
data['Beverage Bottles (Plastic)'] = pd.to_numeric(data['Beverage Bottles (Plastic)'], errors='coerce').fillna(0)

data['Beverage Cans'] = data['Beverage Cans'].astype(str).str.replace(',', '').str.strip()
data['Beverage Cans'] = pd.to_numeric(data['Beverage Cans'], errors='coerce').fillna(0)

# Create scatter plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=data,
    x='Beverage Bottles (Plastic)',
    y='Beverage Cans',
    color='darkorange',
    s=65,
    alpha=0.75
)

# Formatting labels and title
plt.title('Plastic Beverage Bottles vs. Beverage Cans Collected', fontsize=12, fontweight='bold')
plt.xlabel('Plastic Beverage Bottles Count', fontsize=10)
plt.ylabel('Beverage Cans Count', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

This is a good example of a basic but important way to look at data, I would expect beverage cans and plastic bottles to maybe correlate but I wanted to double check, there seems to be a positive trend.
